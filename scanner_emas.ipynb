```python
#Importa las librerias y módulos necesarios.
%load_ext autoreload
%autoreload 2
import pandas as pd
```

```python
import pandas as pd
from datos import obtener_datos_semanales

data_7d = obtener_datos_semanales()
resultados = []

for ticker in data_7d.columns.get_level_values(0).unique():
  df_ticker = data_7d[ticker].dropna(subset=['Close'])
  if len(df_ticker) >= 200:
    cierre = df_ticker['Close']
    ema_150 = cierre.ewm(span=150, adjust=False).mean().iloc[-1].item()
    ema_200 = cierre.ewm(span=200, adjust=False).mean().iloc[-1].item()
    precio = cierre.iloc[-1].item()

    dist_150 = ((precio - ema_150) / ema_150) * 100
    dist_200 = ((precio - ema_200) / ema_200) * 100

    if (abs(dist_150) <= 1.0) or (abs(dist_200) <= 1.0):
      resultados.append({
          'ticker': ticker,
          'precio': precio,
          'dist_150': dist_150,
          'dist_200': dist_200,
      })

for r in resultados:
  print(
      f"Ticker: {r['ticker']} | Precio: {r['precio']:.2f} | Dist. EMA 150:"
      f" {r['dist_150']:.2f}% | Dist. EMA 200: {r['dist_200']:.2f}%"
  )
```
